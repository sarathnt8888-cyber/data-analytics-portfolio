```python
import pandas as pd
```

```python
df=pd.read_csv('HR-Employee-Attrition.xlsx.csv')
```

overall attririon date

```python
overall_rate=(df['Attrition']=='Yes').mean()*100
```

```python
print(f"Overall attrition Rate:{overall_rate:.2f}%\n")
```

attrition by overtime

```python
print("Attrition by OverTime:")
print(df.groupby('OverTime')['Attrition'].apply(lambda x: (x == 'Yes').mean() * 100).round(2))
print()
```

attrition by job role

```python
print("Attrition by Job Role (sorted, highest risk first):")
print(df.groupby('JobRole')['Attrition'].apply(lambda x: (x == 'Yes').mean() * 100)
      .sort_values(ascending=False).round(2))
print()
```

attrition by department

```python
print("Attrition by Department:")
print(df.groupby('Department')['Attrition'].apply(lambda x: (x == 'Yes').mean() * 100)
      .sort_values(ascending=False).round(2))
print()
```

attrition by marital status

```python
print("Attrition by Marital Status:")
print(df.groupby('MaritalStatus')['Attrition'].apply(lambda x: (x == 'Yes').mean() * 100)
      .sort_values(ascending=False).round(2))
print()
```

attrition by business travel

```python
print("Attrition by Business Travel:")
print(df.groupby('BusinessTravel')['Attrition'].apply(lambda x: (x == 'Yes').mean() * 100)
      .sort_values(ascending=False).round(2))
print()
```

attrition by stock option level

```python
print("Attrition by Stock Option Level:")
print(df.groupby('StockOptionLevel')['Attrition'].apply(lambda x: (x == 'Yes').mean() * 100)
      .sort_values(ascending=False).round(2))
print()
```
