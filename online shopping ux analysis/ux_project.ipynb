```python
import pandas as pd
df=pd.read_csv('online_shoppers.csv')
```

```python
print(df.shape)
```

```python

print(df.duplicated().sum())
```

```python
print(df.isna().sum())
```

```python
df=df.drop_duplicates().reset_index(drop=True)
```

```python
df.shape
```

```python
print(df["Month"].unique())
```

```python
df['Month']=df['Month'].replace({'June':'Jun'})
```

```python
print(df["Month"].unique())
```

```python
df['Revenue']=df['Revenue'].astype(int)
```

```python
print(df['Revenue'].mean())
```

```python
monthly=df.groupby('Month')['Revenue'].agg(['count','mean'])
```

```python
print(monthly)
```
